```python
# ============================================================
# BASIC DATA VALIDATION
# ============================================================

print("=== DATA VALIDATION ===")

# Flight count
assert len(flights_25) == 25, "Expected 25 flights"
assert len(aircraft_25) == 8, "Expected 8 aircraft"
assert len(crew_25) == 16, "Expected 16 crew units"

# No flight has same origin and destination
same_airport = flights_25[
    flights_25["origin"] == flights_25["destination"]
]

print("Flights with same origin/destination:")
display(same_airport)

# Arrival must be after departure
invalid_times = flights_25[
    flights_25["sta"] <= flights_25["std"]
]

print("Flights with invalid timing:")
display(invalid_times)

# Forced delay must be non-negative
invalid_delay = flights_25[
    flights_25["forced_delay"] < 0
]

print("Flights with negative forced delay:")
display(invalid_delay)

print("\n=== SUMMARY ===")
print("Flights:", len(flights_25))
print("Aircraft:", len(aircraft_25))
print("Crew:", len(crew_25))

if len(same_airport) == 0 and len(invalid_times) == 0 and len(invalid_delay) == 0:
    print("✓ Basic validation passed")
else:
    print("⚠ Validation found issues")
```

```python
# ============================================================
# PRESOLVER RULE 1:
# AIRCRAFT FIRST-FLIGHT FEASIBILITY
# ============================================================

aircraft_first_results = []

for _, ac in aircraft_25.iterrows():

    for _, flight in flights_25.iterrows():

        reasons = []

        # ----------------------------------------------------
        # Rule 1: Initial location must match flight origin
        # ----------------------------------------------------
        if ac["initial_location"] != flight["origin"]:
            reasons.append("wrong_initial_location")

        # ----------------------------------------------------
        # Rule 2: Aircraft must be available before departure
        # ----------------------------------------------------
        if ac["available_time"] > flight["std"]:
            reasons.append("aircraft_not_available")

        # ----------------------------------------------------
        # Rule 3: Check maintenance / AOG window
        # ----------------------------------------------------
        if pd.notna(ac["unavailability_start"]) and pd.notna(ac["unavailability_end"]):

            # Flight occupies the aircraft from STD to STA.
            # If its operating interval overlaps the
            # unavailability interval, it is infeasible.
            flight_start = flight["std"]
            flight_end = flight["sta"] + flight["forced_delay"]

            unavailable_start = ac["unavailability_start"]
            unavailable_end = ac["unavailability_end"]

            overlap = (
                flight_start < unavailable_end
                and flight_end > unavailable_start
            )

            if overlap:
                reasons.append("technical_or_maintenance_hold")

        feasible = len(reasons) == 0

        aircraft_first_results.append({
            "aircraft_id": ac["aircraft_id"],
            "flight_id": flight["flight_id"],
            "feasible": feasible,
            "reason": ", ".join(reasons)
        })

aircraft_first_df = pd.DataFrame(aircraft_first_results)

print("=== AIRCRAFT FIRST-FLIGHT FEASIBILITY ===")

print("\nFEASIBLE:")
display(
    aircraft_first_df[
        aircraft_first_df["feasible"] == True
    ]
)

print("\nPRUNED:")
display(
    aircraft_first_df[
        aircraft_first_df["feasible"] == False
    ]
)

print("\n=== SUMMARY ===")
print("Total combinations:", len(aircraft_first_df))
print(
    "Feasible:",
    aircraft_first_df["feasible"].sum()
)
print(
    "Pruned:",
    (~aircraft_first_df["feasible"]).sum()
)
```

```python
# ============================================================
# PRESOLVER RULE 2:
# AIRCRAFT TRANSITION FEASIBILITY
# ============================================================

aircraft_transition_results = []

for _, ac in aircraft_25.iterrows():

    for i, f1 in flights_25.iterrows():

        for j, f2 in flights_25.iterrows():

            if i == j:
                continue

            reasons = []

            # ------------------------------------------------
            # Rule 1: Airport continuity
            # ------------------------------------------------
            if f1["destination"] != f2["origin"]:
                reasons.append("airport_mismatch")

            # ------------------------------------------------
            # Rule 2: Calculate actual arrival of F1
            # ------------------------------------------------
            f1_actual_arrival = f1["sta"] + f1["forced_delay"]

            # ------------------------------------------------
            # Rule 3: Calculate available time before F2
            # ------------------------------------------------
            ground_time = f2["std"] - f1_actual_arrival

            if ground_time < ac["turnaround_time_min"]:
                reasons.append("insufficient_turnaround")

            # ------------------------------------------------
            # Rule 4: Check aircraft maintenance/AOG
            # during F2
            # ------------------------------------------------
            if (
                pd.notna(ac["unavailability_start"])
                and pd.notna(ac["unavailability_end"])
            ):

                f2_start = f2["std"]
                f2_end = f2["sta"] + f2["forced_delay"]

                overlap = (
                    f2_start < ac["unavailability_end"]
                    and f2_end > ac["unavailability_start"]
                )

                if overlap:
                    reasons.append("technical_or_maintenance_hold")

            feasible = len(reasons) == 0

            aircraft_transition_results.append({
                "aircraft_id": ac["aircraft_id"],
                "from": f1["flight_id"],
                "to": f2["flight_id"],
                "feasible": feasible,
                "reason": ", ".join(reasons),
                "ground_time": ground_time
            })

aircraft_transition_df = pd.DataFrame(
    aircraft_transition_results
)

print("=== AIRCRAFT TRANSITION FEASIBILITY ===")

print("\nFEASIBLE TRANSITIONS:")
display(
    aircraft_transition_df[
        aircraft_transition_df["feasible"] == True
    ]
)

print("\nPRUNED TRANSITIONS:")
display(
    aircraft_transition_df[
        aircraft_transition_df["feasible"] == False
    ]
)

print("\n=== SUMMARY ===")
print(
    "Total aircraft-flight transitions checked:",
    len(aircraft_transition_df)
)

print(
    "Feasible:",
    aircraft_transition_df["feasible"].sum()
)

print(
    "Pruned:",
    (~aircraft_transition_df["feasible"]).sum()
)
```

```python
# ============================================================
# PRESOLVER RULE 3:
# BUILD FLIGHT-LEVEL ROUTING ARCS
#
# An arc F1 -> F2 exists only if at least one aircraft can:
#   1. Start with F1
#   2. Perform the transition F1 -> F2
#
# Aircraft assignment remains implicit.
# ============================================================

valid_flight_arcs = []

# ------------------------------------------------------------
# Step 1: Get aircraft-flight pairs that are feasible as
#         FIRST flights
# ------------------------------------------------------------

feasible_first = aircraft_first_df[
    aircraft_first_df["feasible"] == True
].copy()

# ------------------------------------------------------------
# Step 2: Get aircraft transitions that are feasible
# ------------------------------------------------------------

feasible_transitions = aircraft_transition_df[
    aircraft_transition_df["feasible"] == True
].copy()

# ------------------------------------------------------------
# Step 3: Match them by aircraft + first flight
# ------------------------------------------------------------

for _, transition in feasible_transitions.iterrows():

    aircraft_id = transition["aircraft_id"]
    from_flight = transition["from"]
    to_flight = transition["to"]

    # Check whether this aircraft can actually START
    # with the first flight.
    can_start = (
        (feasible_first["aircraft_id"] == aircraft_id) &
        (feasible_first["flight_id"] == from_flight)
    ).any()

    if not can_start:
        continue

    valid_flight_arcs.append({
        "from": from_flight,
        "to": to_flight,
        "feasible": True,
        "supporting_aircraft": aircraft_id,
        "ground_time": transition["ground_time"]
    })

# ------------------------------------------------------------
# Step 4: Combine duplicate flight-level arcs
#
# Multiple aircraft may support the same F1 -> F2 arc.
# We keep ONE flight-level arc and list all supporting aircraft.
# ------------------------------------------------------------

if valid_flight_arcs:

    flight_arc_temp_df = pd.DataFrame(valid_flight_arcs)

    flight_arc_df = (
        flight_arc_temp_df
        .groupby(["from", "to"], as_index=False)
        .agg(
            feasible=("feasible", "max"),
            supporting_aircraft=(
                "supporting_aircraft",
                lambda x: ", ".join(sorted(x.unique()))
            ),
            aircraft_count=(
                "supporting_aircraft",
                "nunique"
            ),
            ground_time=("ground_time", "first")
        )
    )

else:

    flight_arc_df = pd.DataFrame(
        columns=[
            "from",
            "to",
            "feasible",
            "supporting_aircraft",
            "aircraft_count",
            "ground_time"
        ]
    )

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("=== VALID FLIGHT-LEVEL ROUTING ARCS ===")

display(
    flight_arc_df
    .sort_values(["from", "to"])
    .reset_index(drop=True)
)

print("\n=== SUMMARY ===")
print(
    "Valid flight-level arcs:",
    len(flight_arc_df)
)

print(
    "Aircraft-supported transitions used:",
    len(valid_flight_arcs)
)
```

```python
# ============================================================
# PRESOLVER RULE 4:
# CREW FIRST-FLIGHT FEASIBILITY
# ============================================================

crew_first_results = []

for _, cr in crew_25.iterrows():

    for _, flight in flights_25.iterrows():

        reasons = []

        # ----------------------------------------------------
        # Rule 1: Crew home base must match flight origin
        # ----------------------------------------------------
        if cr["home_base"] != flight["origin"]:
            reasons.append("wrong_home_base")

        # ----------------------------------------------------
        # Rule 2: Crew must be available before departure
        # ----------------------------------------------------
        if cr["available_time"] > flight["std"]:
            reasons.append("crew_not_available")

        # ----------------------------------------------------
        # Rule 3: Flight duration must not exceed
        # maximum allowed flight time
        # ----------------------------------------------------
        flight_duration = (
            flight["sta"] - flight["std"]
        )

        if flight_duration > cr["max_flight_mins"]:
            reasons.append("exceeds_max_flight_time")

        feasible = len(reasons) == 0

        crew_first_results.append({
            "crew_id": cr["crew_id"],
            "flight_id": flight["flight_id"],
            "feasible": feasible,
            "reason": ", ".join(reasons)
        })

crew_first_df = pd.DataFrame(crew_first_results)

print("=== CREW FIRST-FLIGHT FEASIBILITY ===")

print("\nFEASIBLE:")
display(
    crew_first_df[
        crew_first_df["feasible"] == True
    ]
)

print("\nPRUNED:")
display(
    crew_first_df[
        crew_first_df["feasible"] == False
    ]
)

print("\n=== SUMMARY ===")
print(
    "Total combinations:",
    len(crew_first_df)
)

print(
    "Feasible:",
    crew_first_df["feasible"].sum()
)

print(
    "Pruned:",
    (~crew_first_df["feasible"]).sum()
)
```

```python
# ============================================================
# PRESOLVER RULE 5:
# CREW TRANSITION FEASIBILITY
# ============================================================

crew_transition_results = []

for _, cr in crew_25.iterrows():

    for _, f1 in flights_25.iterrows():

        for _, f2 in flights_25.iterrows():

            if f1["flight_id"] == f2["flight_id"]:
                continue

            reasons = []

            # ------------------------------------------------
            # Rule 1: Crew must be able to operate F1
            # as a first flight
            # ------------------------------------------------
            f1_first = crew_first_df[
                (crew_first_df["crew_id"] == cr["crew_id"]) &
                (crew_first_df["flight_id"] == f1["flight_id"]) &
                (crew_first_df["feasible"] == True)
            ]

            if f1_first.empty:
                reasons.append("cannot_start_first_flight")

            # ------------------------------------------------
            # Rule 2: Flight continuity
            # Crew can only continue if the next flight
            # starts from the previous flight's destination.
            # ------------------------------------------------
            if f1["destination"] != f2["origin"]:
                reasons.append("airport_mismatch")

            # ------------------------------------------------
            # Rule 3: Rest between flights
            # ------------------------------------------------
            f1_actual_arrival = (
                f1["sta"] + f1["forced_delay"]
            )

            ground_time = (
                f2["std"] - f1_actual_arrival
            )

            if ground_time < cr["min_rest_mins"]:
                reasons.append("insufficient_rest")

            # ------------------------------------------------
            # Rule 4: Total duty time
            # From beginning of F1 to end of F2
            # ------------------------------------------------
            duty_time = (
                f2["sta"] + f2["forced_delay"]
                - f1["std"]
            )

            if duty_time > cr["max_duty_mins"]:
                reasons.append("exceeds_max_duty_time")

            feasible = len(reasons) == 0

            crew_transition_results.append({
                "crew_id": cr["crew_id"],
                "from": f1["flight_id"],
                "to": f2["flight_id"],
                "feasible": feasible,
                "reason": ", ".join(reasons),
                "ground_time": ground_time,
                "duty_time": duty_time
            })

crew_transition_df = pd.DataFrame(
    crew_transition_results
)

print("=== CREW TRANSITION FEASIBILITY ===")

print("\nFEASIBLE TRANSITIONS:")
display(
    crew_transition_df[
        crew_transition_df["feasible"] == True
    ]
)

print("\nPRUNED TRANSITIONS:")
display(
    crew_transition_df[
        crew_transition_df["feasible"] == False
    ]
)

print("\n=== SUMMARY ===")
print(
    "Total crew-flight transitions checked:",
    len(crew_transition_df)
)

print(
    "Feasible:",
    crew_transition_df["feasible"].sum()
)

print(
    "Pruned:",
    (~crew_transition_df["feasible"]).sum()
)
```

```python
# ============================================================
# PRESOLVER RULE 6:
# FINAL FLIGHT ROUTING FEASIBILITY
#
# A flight transition is valid only when:
#   1. At least one aircraft can perform it
#   2. At least one crew member can perform it
#
# Aircraft and crew assignments remain implicit.
# ============================================================

# ------------------------------------------------------------
# Aircraft-feasible transitions
# ------------------------------------------------------------

aircraft_feasible = aircraft_transition_df[
    aircraft_transition_df["feasible"] == True
].copy()

# ------------------------------------------------------------
# Crew-feasible transitions
# ------------------------------------------------------------

crew_feasible = crew_transition_df[
    crew_transition_df["feasible"] == True
].copy()

# ------------------------------------------------------------
# Get unique flight pairs for each resource
# ------------------------------------------------------------

aircraft_arcs = set(
    zip(
        aircraft_feasible["from"],
        aircraft_feasible["to"]
    )
)

crew_arcs = set(
    zip(
        crew_feasible["from"],
        crew_feasible["to"]
    )
)

# ------------------------------------------------------------
# Final intersection
# ------------------------------------------------------------

final_arcs = sorted(
    aircraft_arcs.intersection(crew_arcs)
)

# ------------------------------------------------------------
# Build final routing table
# ------------------------------------------------------------

final_routing_results = []

for from_flight, to_flight in final_arcs:

    # Aircraft supporting this transition
    supporting_aircraft = sorted(
        aircraft_feasible[
            (aircraft_feasible["from"] == from_flight) &
            (aircraft_feasible["to"] == to_flight)
        ]["aircraft_id"].unique()
    )

    # Crew supporting this transition
    supporting_crew = sorted(
        crew_feasible[
            (crew_feasible["from"] == from_flight) &
            (crew_feasible["to"] == to_flight)
        ]["crew_id"].unique()
    )

    # Ground time
    ground_time = aircraft_feasible[
        (aircraft_feasible["from"] == from_flight) &
        (aircraft_feasible["to"] == to_flight)
    ]["ground_time"].iloc[0]

    final_routing_results.append({
        "from": from_flight,
        "to": to_flight,
        "feasible": True,
        "supporting_aircraft": ", ".join(supporting_aircraft),
        "supporting_crew": ", ".join(supporting_crew),
        "aircraft_count": len(supporting_aircraft),
        "crew_count": len(supporting_crew),
        "ground_time": ground_time
    })

final_routing_df = pd.DataFrame(
    final_routing_results
)

# ------------------------------------------------------------
# Display final result
# ------------------------------------------------------------

print("=== FINAL FEASIBLE ROUTING ARCS ===")

display(
    final_routing_df
    .sort_values(["from", "to"])
    .reset_index(drop=True)
)

print("\n=== SUMMARY ===")

print(
    "Aircraft-feasible flight arcs:",
    len(aircraft_arcs)
)

print(
    "Crew-feasible flight arcs:",
    len(crew_arcs)
)

print(
    "Final feasible routing arcs:",
    len(final_arcs)
)
```

```python
# ============================================================
# MILP MODEL: DECISION VARIABLES
# ============================================================

from scipy.optimize import milp, LinearConstraint, Bounds
from scipy.sparse import lil_matrix
import numpy as np

# ------------------------------------------------------------
# 1. Cancellation variables
# ------------------------------------------------------------

cancel_vars = [
    f"x_cancel_{flight_id}"
    for flight_id in flights_25["flight_id"]
]

# ------------------------------------------------------------
# 2. Delay variables
# ------------------------------------------------------------

delay_vars = [
    f"d_{flight_id}"
    for flight_id in flights_25["flight_id"]
]

# ------------------------------------------------------------
# 3. Routing arc variables
#
# final_arcs contains tuples:
# ('F01', 'F19')
# ('F02', 'F03')
# etc.
# ------------------------------------------------------------

arc_vars = [
    f"x_arc_{from_flight}_{to_flight}"
    for from_flight, to_flight in final_arcs
]

# ------------------------------------------------------------
# 4. Combine all variables
# ------------------------------------------------------------

variable_names = (
    cancel_vars
    + delay_vars
    + arc_vars
)

# ------------------------------------------------------------
# 5. Variable types
#
# B = Binary
# C = Continuous
# ------------------------------------------------------------

variable_types = (
    ["B"] * len(cancel_vars)
    + ["C"] * len(delay_vars)
    + ["B"] * len(arc_vars)
)

# ------------------------------------------------------------
# 6. Lower bounds
# ------------------------------------------------------------

lower_bounds = (
    [0.0] * len(cancel_vars)
    + [0.0] * len(delay_vars)
    + [0.0] * len(arc_vars)
)

# ------------------------------------------------------------
# 7. Upper bounds
#
# Cancellation: 0 or 1
# Delay: maximum 60 minutes
# Routing arc: 0 or 1
# ------------------------------------------------------------

upper_bounds = (
    [1.0] * len(cancel_vars)
    + [60.0] * len(delay_vars)
    + [1.0] * len(arc_vars)
)

# ------------------------------------------------------------
# 8. Display model size
# ------------------------------------------------------------

print("=== MILP DECISION VARIABLES ===")

print(f"Cancellation variables : {len(cancel_vars)}")
print(f"Delay variables        : {len(delay_vars)}")
print(f"Routing arc variables  : {len(arc_vars)}")
print(f"Total decision variables: {len(variable_names)}")

# ------------------------------------------------------------
# 9. Display variable list
# ------------------------------------------------------------

print("\n=== VARIABLE LIST ===")

for i, (name, vtype, lb, ub) in enumerate(
    zip(
        variable_names,
        variable_types,
        lower_bounds,
        upper_bounds
    )
):
    print(
        f"{i:3d} | {name:25s} | "
        f"Type: {vtype} | LB: {lb} | UB: {ub}"
    )
```

```python
# ============================================================
# MILP OBJECTIVE FUNCTION
# ============================================================

# Objective coefficient for every decision variable
objective = np.zeros(len(variable_names))

# ------------------------------------------------------------
# 1. Cancellation costs
# ------------------------------------------------------------

for i, flight_id in enumerate(flights_25["flight_id"]):

    flight = flights_25[
        flights_25["flight_id"] == flight_id
    ].iloc[0]

    objective[i] = flight["c_cancel"]


# ------------------------------------------------------------
# 2. Delay costs
# ------------------------------------------------------------

delay_start = len(cancel_vars)

for i, flight_id in enumerate(flights_25["flight_id"]):

    flight = flights_25[
        flights_25["flight_id"] == flight_id
    ].iloc[0]

    objective[delay_start + i] = flight["c_delay_per_min"]


# ------------------------------------------------------------
# 3. Routing arc costs
# ------------------------------------------------------------

arc_start = len(cancel_vars) + len(delay_vars)

for i, (_, arc) in enumerate(final_routing_df.iterrows()):

    ground_time = arc["ground_time"]

    objective[arc_start + i] = ground_time * 500


# ------------------------------------------------------------
# Display objective
# ------------------------------------------------------------

print("=== OBJECTIVE FUNCTION ===")

for i, (name, cost) in enumerate(
    zip(variable_names, objective)
):

    print(
        f"{i:3d} | {name:25s} | "
        f"Cost coefficient: ₹{cost:,.0f}"
    )

print("\n=== OBJECTIVE SUMMARY ===")

print(
    f"Cancellation cost coefficients: "
    f"{objective[:len(cancel_vars)].min():,.0f} "
    f"to "
    f"{objective[:len(cancel_vars)].max():,.0f}"
)

print(
    f"Delay cost per minute: "
    f"{objective[delay_start]:,.0f}"
)

print(
    f"Routing variables: "
    f"{len(arc_vars)}"
)
```

```python
# ============================================================
# FLIGHT COVERAGE CONSTRAINTS
# ============================================================

print("=== ADDING FLIGHT COVERAGE CONSTRAINTS ===")

# ------------------------------------------------------------
# 1. Create variable-index mapping
#
# This lets us find the position of every MILP variable
# inside the objective / constraint matrix.
# ------------------------------------------------------------

var_index = {
    name: i
    for i, name in enumerate(variable_names)
}

# ------------------------------------------------------------
# 2. Prepare constraint matrix
#
# One coverage constraint for every flight.
# ------------------------------------------------------------

n_flights = len(flights_25)
n_variables = len(variable_names)

coverage_matrix = lil_matrix(
    (n_flights, n_variables),
    dtype=float
)

coverage_lower = np.ones(n_flights)
coverage_upper = np.full(n_flights, np.inf)

# ------------------------------------------------------------
# 3. Build one constraint for each flight
#
# Flight must either:
#
#   - be cancelled
#   OR
#   - have an incoming routing arc
#   OR
#   - have an outgoing routing arc
#
# Constraint:
#
# x_cancel_Fi
# + incoming arcs
# + outgoing arcs >= 1
# ------------------------------------------------------------

coverage_constraints = []

for row_idx, flight_id in enumerate(
    flights_25["flight_id"]
):

    # --------------------------------------------------------
    # Cancellation variable
    # --------------------------------------------------------

    cancel_name = f"x_cancel_{flight_id}"

    coverage_matrix[
        row_idx,
        var_index[cancel_name]
    ] = 1

    # --------------------------------------------------------
    # Find incoming and outgoing arcs
    #
    # final_arcs contains tuples:
    # ('F01', 'F19')
    # ('F02', 'F03')
    # etc.
    # --------------------------------------------------------

    incoming_arcs = [
        arc
        for arc in final_arcs
        if arc[1] == flight_id
    ]

    outgoing_arcs = [
        arc
        for arc in final_arcs
        if arc[0] == flight_id
    ]

    # --------------------------------------------------------
    # Add incoming arc variables
    # --------------------------------------------------------

    for from_flight, to_flight in incoming_arcs:

        arc_name = (
            f"x_arc_{from_flight}_{to_flight}"
        )

        coverage_matrix[
            row_idx,
            var_index[arc_name]
        ] = 1

    # --------------------------------------------------------
    # Add outgoing arc variables
    # --------------------------------------------------------

    for from_flight, to_flight in outgoing_arcs:

        arc_name = (
            f"x_arc_{from_flight}_{to_flight}"
        )

        coverage_matrix[
            row_idx,
            var_index[arc_name]
        ] = 1

    # --------------------------------------------------------
    # Store information for validation
    # --------------------------------------------------------

    coverage_constraints.append({
        "flight_id": flight_id,
        "incoming_arcs": incoming_arcs,
        "outgoing_arcs": outgoing_arcs
    })


# ------------------------------------------------------------
# 4. Convert sparse matrix to CSR
# ------------------------------------------------------------

coverage_matrix = coverage_matrix.tocsr()

# ------------------------------------------------------------
# 5. Create SciPy LinearConstraint
# ------------------------------------------------------------

coverage_constraint = LinearConstraint(
    coverage_matrix,
    coverage_lower,
    coverage_upper
)

# ------------------------------------------------------------
# 6. Display summary
# ------------------------------------------------------------

print(
    f"Coverage constraints added: "
    f"{len(coverage_constraints)}"
)

print(
    f"Expected: "
    f"{len(flights_25)}"
)

# ------------------------------------------------------------
# 7. Display constraint structure
# ------------------------------------------------------------

print("\n=== COVERAGE CONSTRAINT DETAILS ===")

for item in coverage_constraints:

    print(
        f"{item['flight_id']} | "
        f"Incoming: {len(item['incoming_arcs'])} | "
        f"Outgoing: {len(item['outgoing_arcs'])}"
    )

print("\n=== COVERAGE MATRIX ===")
print(
    f"Rows (constraints): {coverage_matrix.shape[0]}"
)

print(
    f"Columns (variables): {coverage_matrix.shape[1]}"
)

print(
    f"Non-zero coefficients: "
    f"{coverage_matrix.nnz}"
)
```

```python
# ============================================================
# FORCED DELAY CONSTRAINTS
# ============================================================

print("=== ADDING FORCED DELAY CONSTRAINTS ===")

# ------------------------------------------------------------
# 1. Create constraint matrix
#
# One constraint for every flight:
#
#       d_Fxx >= forced_delay
# ------------------------------------------------------------

n_flights = len(flights_25)
n_variables = len(variable_names)

delay_constraint_matrix = lil_matrix(
    (n_flights, n_variables),
    dtype=float
)

delay_lower = np.zeros(n_flights)
delay_upper = np.full(n_flights, np.inf)

# ------------------------------------------------------------
# 2. Add forced-delay constraints
# ------------------------------------------------------------

forced_delay_constraints = []

for row_idx, flight_id in enumerate(
    flights_25["flight_id"]
):

    # Variable name
    delay_name = f"d_{flight_id}"

    # Column corresponding to delay variable
    delay_idx = var_index[delay_name]

    # Coefficient of delay variable = 1
    delay_constraint_matrix[
        row_idx,
        delay_idx
    ] = 1

    # Get forced delay from dataset
    flight = flights_25[
        flights_25["flight_id"] == flight_id
    ].iloc[0]

    forced_delay = float(
        flight["forced_delay"]
    )

    # Constraint:
    #
    # d_Fxx >= forced_delay
    #
    delay_lower[row_idx] = forced_delay

    forced_delay_constraints.append({
        "flight_id": flight_id,
        "forced_delay": forced_delay,
        "variable": delay_name
    })

# ------------------------------------------------------------
# 3. Convert to CSR format
# ------------------------------------------------------------

delay_constraint_matrix = (
    delay_constraint_matrix.tocsr()
)

# ------------------------------------------------------------
# 4. Create LinearConstraint
# ------------------------------------------------------------

delay_constraint = LinearConstraint(
    delay_constraint_matrix,
    delay_lower,
    delay_upper
)

# ------------------------------------------------------------
# 5. Display results
# ------------------------------------------------------------

print(
    f"Forced delay constraints added: "
    f"{len(forced_delay_constraints)}"
)

print(
    f"Expected: {len(flights_25)}"
)

print("\n=== FORCED DELAY CONSTRAINT DETAILS ===")

for item in forced_delay_constraints:

    print(
        f"{item['flight_id']} | "
        f"{item['variable']} >= "
        f"{item['forced_delay']:.0f} min"
    )

print("\n=== DELAY CONSTRAINT MATRIX ===")

print(
    f"Rows (constraints): "
    f"{delay_constraint_matrix.shape[0]}"
)

print(
    f"Columns (variables): "
    f"{delay_constraint_matrix.shape[1]}"
)

print(
    f"Non-zero coefficients: "
    f"{delay_constraint_matrix.nnz}"
)
```

```python
# ============================================================
# UPDATE DELAY VARIABLE UPPER BOUND
# ============================================================

MAX_DELAY = 180.0

# Delay variables occupy the positions immediately
# after the cancellation variables.

delay_start = len(cancel_vars)
delay_end = delay_start + len(delay_vars)

for i in range(delay_start, delay_end):
    upper_bounds[i] = MAX_DELAY

print("=== UPDATED DELAY BOUNDS ===")
print(f"Maximum allowed additional delay: {MAX_DELAY:.0f} minutes")

print("\nDelay variable bounds:")

for i, flight_id in enumerate(flights_25["flight_id"]):

    idx = delay_start + i

    print(
        f"d_{flight_id:3s} | "
        f"LB: {lower_bounds[idx]:.0f} | "
        f"UB: {upper_bounds[idx]:.0f}"
    )
```

```python
# ============================================================
# ROUTING CONTINUITY CONSTRAINTS
# ============================================================

print("=== ADDING ROUTING CONTINUITY CONSTRAINTS ===")

continuity_constraints = []

# ------------------------------------------------------------
# Create a mapping:
# variable name -> column index
# ------------------------------------------------------------

variable_index = {
    name: i
    for i, name in enumerate(variable_names)
}

# ------------------------------------------------------------
# Arc variables start after cancellation + delay variables
# ------------------------------------------------------------

arc_start = len(cancel_vars) + len(delay_vars)

# ------------------------------------------------------------
# Build continuity constraints for every flight
#
# For each flight:
#
# Incoming arcs - Outgoing arcs = 0
#
# This ensures that if a flight is connected to another
# flight before it, the routing flow is consistent.
# ------------------------------------------------------------

for flight_id in flights_25["flight_id"]:

    # Find incoming arcs
    incoming_arcs = [
        arc for arc in final_arcs
        if arc[1] == flight_id
    ]

    # Find outgoing arcs
    outgoing_arcs = [
        arc for arc in final_arcs
        if arc[0] == flight_id
    ]

    # --------------------------------------------------------
    # Build coefficient dictionary for this constraint
    # --------------------------------------------------------

    coefficients = {}

    # Incoming arcs get +1
    for from_flight, to_flight in incoming_arcs:

        arc_name = f"x_arc_{from_flight}_{to_flight}"
        col_index = variable_index[arc_name]

        coefficients[col_index] = (
            coefficients.get(col_index, 0) + 1
        )

    # Outgoing arcs get -1
    for from_flight, to_flight in outgoing_arcs:

        arc_name = f"x_arc_{from_flight}_{to_flight}"
        col_index = variable_index[arc_name]

        coefficients[col_index] = (
            coefficients.get(col_index, 0) - 1
        )

    # --------------------------------------------------------
    # Store constraint information
    # --------------------------------------------------------

    continuity_constraints.append({
        "flight_id": flight_id,
        "coefficients": coefficients,
        "incoming_count": len(incoming_arcs),
        "outgoing_count": len(outgoing_arcs)
    })


# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print(
    f"Routing continuity constraints prepared: "
    f"{len(continuity_constraints)}"
)

print(
    f"Expected: {len(flights_25)}"
)

print("\n=== ROUTING CONTINUITY DETAILS ===")

for item in continuity_constraints:

    print(
        f"{item['flight_id']} | "
        f"Incoming: {item['incoming_count']} | "
        f"Outgoing: {item['outgoing_count']}"
    )


# ============================================================
# BUILD CONTINUITY MATRIX
# ============================================================

n_constraints = len(continuity_constraints)
n_variables = len(variable_names)

continuity_matrix = lil_matrix(
    (n_constraints, n_variables),
    dtype=float
)

for row, item in enumerate(continuity_constraints):

    for col, coefficient in item["coefficients"].items():

        continuity_matrix[row, col] = coefficient


continuity_matrix = continuity_matrix.tocsr()


# ------------------------------------------------------------
# Equality bounds
#
# Incoming - Outgoing = 0
# ------------------------------------------------------------

continuity_lb = np.zeros(n_constraints)
continuity_ub = np.zeros(n_constraints)


print("\n=== ROUTING CONTINUITY MATRIX ===")

print(
    f"Rows (constraints): "
    f"{continuity_matrix.shape[0]}"
)

print(
    f"Columns (variables): "
    f"{continuity_matrix.shape[1]}"
)

print(
    f"Non-zero coefficients: "
    f"{continuity_matrix.nnz}"
)
```

```python
# ============================================================
# FLIGHT CANCELLATION - ROUTING CONSISTENCY CONSTRAINTS
# ============================================================

print("=== ADDING CANCELLATION-ROUTING CONSISTENCY CONSTRAINTS ===")

cancellation_routing_constraints = []

# Map flight ID -> cancellation variable index
cancel_index = {
    flight_id: i
    for i, flight_id in enumerate(flights_25["flight_id"])
}

# Starting index of routing variables
arc_start = len(cancel_vars) + len(delay_vars)

for j, (from_flight, to_flight) in enumerate(final_arcs):

    arc_var_index = arc_start + j

    from_cancel_index = cancel_index[from_flight]
    to_cancel_index = cancel_index[to_flight]

    # x_arc + x_cancel_from <= 1
    row = np.zeros(len(variable_names))

    row[arc_var_index] = 1
    row[from_cancel_index] = 1

    cancellation_routing_constraints.append(
        (
            from_flight,
            to_flight,
            "from_cancel",
            row
        )
    )

    # x_arc + x_cancel_to <= 1
    row = np.zeros(len(variable_names))

    row[arc_var_index] = 1
    row[to_cancel_index] = 1

    cancellation_routing_constraints.append(
        (
            from_flight,
            to_flight,
            "to_cancel",
            row
        )
    )


print(
    "Cancellation-routing constraints prepared:",
    len(cancellation_routing_constraints)
)

print(
    "Expected:",
    2 * len(final_arcs)
)

# ------------------------------------------------------------
# Display details
# ------------------------------------------------------------

print("\n=== CANCELLATION-ROUTING DETAILS ===")

for from_flight, to_flight, direction, row in (
    cancellation_routing_constraints
):

    if direction == "from_cancel":
        print(
            f"{from_flight} -> {to_flight} | "
            f"x_arc + x_cancel_{from_flight} <= 1"
        )

    else:
        print(
            f"{from_flight} -> {to_flight} | "
            f"x_arc + x_cancel_{to_flight} <= 1"
        )


# ------------------------------------------------------------
# Build matrix
# ------------------------------------------------------------

if cancellation_routing_constraints:

    cancellation_routing_matrix = np.vstack(
        [
            row
            for _, _, _, row
            in cancellation_routing_constraints
        ]
    )

else:

    cancellation_routing_matrix = np.zeros(
        (0, len(variable_names))
    )


print("\n=== CANCELLATION-ROUTING MATRIX ===")

print(
    "Rows (constraints):",
    cancellation_routing_matrix.shape[0]
)

print(
    "Columns (variables):",
    cancellation_routing_matrix.shape[1]
)

print(
    "Non-zero coefficients:",
    np.count_nonzero(cancellation_routing_matrix)
)
```

```python
# ============================================================
# DELAY - ROUTING FEASIBILITY CONSTRAINTS
# ============================================================

print("=== ADDING DELAY-ROUTING CONSTRAINTS ===")

delay_routing_constraints = []

# ------------------------------------------------------------
# 1. Map each flight to its delay-variable index
# ------------------------------------------------------------

delay_index = {
    flight_id: len(cancel_vars) + i
    for i, flight_id in enumerate(flights_25["flight_id"])
}

# ------------------------------------------------------------
# 2. Starting index of routing variables
# ------------------------------------------------------------

arc_start = len(cancel_vars) + len(delay_vars)

# ------------------------------------------------------------
# 3. Minimum turnaround time
# ------------------------------------------------------------

MIN_REST = 30

# Big-M for conditional routing constraint
BIG_M = 10000

# ------------------------------------------------------------
# 4. Build one constraint for every routing arc
# ------------------------------------------------------------

for j, (from_flight, to_flight) in enumerate(final_arcs):

    # --------------------------------------------------------
    # Get flight information
    # --------------------------------------------------------

    from_row = flights_25[
        flights_25["flight_id"] == from_flight
    ].iloc[0]

    to_row = flights_25[
        flights_25["flight_id"] == to_flight
    ].iloc[0]

    # Scheduled arrival of first flight
    arrival_time = from_row["sta"]

    # Scheduled departure of second flight
    departure_time = to_row["std"]

    # Scheduled ground time
    ground_time = departure_time - arrival_time

    # --------------------------------------------------------
    # Variable indices
    # --------------------------------------------------------

    from_delay_index = delay_index[from_flight]
    to_delay_index = delay_index[to_flight]

    arc_var_index = arc_start + j

    # --------------------------------------------------------
    # Constraint:
    #
    # STA(F1) + d(F1) + MIN_REST
    #       <=
    # STD(F2) + d(F2) + M(1 - x_arc)
    #
    # Rearranged:
    #
    # d(F1) - d(F2) + M*x_arc
    #       <=
    # ground_time - MIN_REST + M
    # --------------------------------------------------------

    row = np.zeros(len(variable_names))

    row[from_delay_index] = 1
    row[to_delay_index] = -1
    row[arc_var_index] = BIG_M

    rhs = (
        ground_time
        - MIN_REST
        + BIG_M
    )

    delay_routing_constraints.append(
        {
            "from": from_flight,
            "to": to_flight,
            "ground_time": ground_time,
            "rhs": rhs,
            "row": row
        }
    )


# ------------------------------------------------------------
# 5. Display summary
# ------------------------------------------------------------

print(
    "Delay-routing constraints prepared:",
    len(delay_routing_constraints)
)

print(
    "Expected:",
    len(final_arcs)
)


# ------------------------------------------------------------
# 6. Display constraint details
# ------------------------------------------------------------

print("\n=== DELAY-ROUTING DETAILS ===")

for constraint in delay_routing_constraints:

    print(
        f"{constraint['from']} -> {constraint['to']} | "
        f"Scheduled ground time: "
        f"{constraint['ground_time']} min | "
        f"Minimum turnaround: {MIN_REST} min"
    )


# ------------------------------------------------------------
# 7. Build constraint matrix
# ------------------------------------------------------------

if delay_routing_constraints:

    delay_routing_matrix = np.vstack(
        [
            constraint["row"]
            for constraint in delay_routing_constraints
        ]
    )

    delay_routing_rhs = np.array(
        [
            constraint["rhs"]
            for constraint in delay_routing_constraints
        ]
    )

else:

    delay_routing_matrix = np.zeros(
        (0, len(variable_names))
    )

    delay_routing_rhs = np.array([])


# ------------------------------------------------------------
# 8. Display matrix information
# ------------------------------------------------------------

print("\n=== DELAY-ROUTING MATRIX ===")

print(
    "Rows (constraints):",
    delay_routing_matrix.shape[0]
)

print(
    "Columns (variables):",
    delay_routing_matrix.shape[1]
)

print(
    "Non-zero coefficients:",
    np.count_nonzero(delay_routing_matrix)
)
```

```python
# ============================================================
# COMBINE ALL MILP CONSTRAINTS
# ============================================================

print("=== COMBINING MILP CONSTRAINTS ===")

# ------------------------------------------------------------
# 1. Convert all matrices to compatible sparse format
# ------------------------------------------------------------

from scipy.sparse import vstack, csr_matrix

coverage_matrix_sparse = csr_matrix(coverage_matrix)
delay_matrix_sparse = csr_matrix(delay_constraint_matrix)
continuity_matrix_sparse = csr_matrix(continuity_matrix)
delay_routing_matrix_sparse = csr_matrix(delay_routing_matrix)


# ------------------------------------------------------------
# 2. Combine all constraint matrices
# ------------------------------------------------------------

constraint_matrix = vstack([
    coverage_matrix_sparse,
    delay_matrix_sparse,
    continuity_matrix_sparse,
    delay_routing_matrix_sparse
]).tocsr()


# ------------------------------------------------------------
# 3. Combine lower bounds
# ------------------------------------------------------------

delay_routing_lower = np.full(
    len(delay_routing_constraints),
    -np.inf
)

constraint_lower = np.concatenate([
    coverage_lower,
    delay_lower,
    continuity_lb,
    delay_routing_lower
])


# ------------------------------------------------------------
# 4. Combine upper bounds
# ------------------------------------------------------------

constraint_upper = np.concatenate([
    coverage_upper,
    delay_upper,
    continuity_ub,
    delay_routing_rhs
])


# ------------------------------------------------------------
# 5. Create final LinearConstraint object
# ------------------------------------------------------------

all_constraints = LinearConstraint(
    constraint_matrix,
    constraint_lower,
    constraint_upper
)


# ------------------------------------------------------------
# 6. Display final constraint information
# ------------------------------------------------------------

print("\n=== FINAL CONSTRAINT MATRIX ===")

print(
    "Rows (total constraints):",
    constraint_matrix.shape[0]
)

print(
    "Columns (decision variables):",
    constraint_matrix.shape[1]
)

print(
    "Non-zero coefficients:",
    constraint_matrix.nnz
)


print("\n=== CONSTRAINT BREAKDOWN ===")

print(
    "Flight coverage constraints    :",
    coverage_matrix.shape[0]
)

print(
    "Forced delay constraints       :",
    delay_constraint_matrix.shape[0]
)

print(
    "Routing continuity constraints :",
    continuity_matrix.shape[0]
)

print(
    "Delay-routing constraints      :",
    delay_routing_matrix.shape[0]
)

print(
    "---------------------------------------"
)

print(
    "Total constraints              :",
    constraint_matrix.shape[0]
)
```
